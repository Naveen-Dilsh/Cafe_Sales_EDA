### 1. Import Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### 2. Basic Preprocessing

```python
df = pd.read_csv("data/processed/Cafe_sales_handled_outliers.csv")
df.head()
```

```python
Categorical_Features = ["Item","Payment Method","Location"]

Nominal_Features = ["Item","Payment Method","Location"]
Ordinal_Features = [] # There is no ordinal features because those features does not contain a specific order
```

### 3 . Encode Nominal Features

```python
Item_dummies = pd.get_dummies(df["Item"], prefix="Item", dtype=int)
Payment_Method_dummies = pd.get_dummies(df["Payment Method"], prefix="Payment", dtype=int)
Location_dummies = pd.get_dummies(df["Location"], prefix="Location", dtype=int)

df_encoded = pd.concat([df,Item_dummies ],axis = 1)
del df_encoded["Item"]

df_encoded = pd.concat([df_encoded,Payment_Method_dummies ],axis = 1)
del df_encoded["Payment Method"]

df_encoded = pd.concat([df_encoded,Location_dummies ],axis = 1)
del df_encoded["Location"]

df_encoded
```
