### 1. important Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

### 2. Basic Processing

```python
df = pd.read_csv("data/processed/Cafe_sales_feature_encoded.csv")
df.head()
```

### No need to scale this dataset because
- Very Limited Range & Discrete Values
- Already Similar Scales
- Scaling Discrete Data Issues
    - Scaling works best with continuous data  
    - With only 5 unique values, scaling might create artificial precision             

- But i am scaling because of learning Purposes. After looking model accuracy i will change that according to the model

```python
columns_need_to_be_scaled = ["Quantity","Price Per Unit","Total Spent"]

for col in columns_need_to_be_scaled:
    scaler = StandardScaler()
    df[col] =scaler.fit_transform(df[[col]])
```
