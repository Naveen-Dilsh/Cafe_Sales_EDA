**Feature Binning = Numerical Only**

- What binning does:

    - Takes continuous/numerical values and groups them into discrete ranges
        - Example: Ages 25, 26, 27, 28, 29 → "20-30" bin

- Categorical Cannot be binned because:

    - No numerical order: "Coffee" isn't greater than "Cake"
    - No meaningful ranges: You can't create ranges like "Coffee-Cake"
    - Already discrete: They're already in their simplest categorical form

### There is no need to feature Binning in here. because all numerical column already naturally discrete
- Price Per Unit: Only 6 unique values - already naturally discrete
- Total Spent: Only 5 unique values - already naturally discrete
